```python
#  Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

x1, y1 = map(int,input("enter first x and y coordinet").split())
x2, y2 = map(int,input(" enter seconde x and y coordinet").split())

import math
eucli = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
print (f"ans is {eucli}")
```

```python
# Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.

p = int(input("enter principle amount"))
r = int(input("enter rate of intrest"))
t = int(input("enter time"))

si = (p*r*t)/ 100
print(f"simple intrest is {si}")
```

```python
#  Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.
h = int(input("enter the head"))
l = int(input("enter the legs"))

d = (l - (2*h)) / 2
c = h - d

print(f"total no of dogs{d}")
print(f"total no of chicken{c}")
```

```python
#  Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

num = int(input("enter the number"))
sumofnum = 0
for i in range(num+1):
    sumofnum = sumofnum + i

print(sumofnum)
```

```python
# Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.
#  volumn of the tank Tank Volume=Height⋅Width⋅Breadth
# volumn of the glass Volume=𝜋⋅𝑟2⋅ℎ

h,w,b = map(int,input("enter the height, width and breadth of the tank").split())
r,H = map(int,input("enter the radius and height of the glass").split())

tank = h*w*b
glass = (22/7 * r **2) * H

noOfGlass = tank / glass

print(f"No of milk glasses are in tank = {noOfGlass}")
```
