```python
# Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.
#  volumn of the tank Tank Volume=Height⋅Width⋅Breadth
# volumn of the glass Volume=𝜋⋅𝑟2⋅ℎ

h,w,b = map(int,input("enter the height, width and breadth of the tank").split())
r,H = map(int,input("enter the radius and height of the glass").split())

tank = h*w*b
glass = (22/7 * r **2) * H

noOfGlass = tank / glass

print(f"No of milk glasses are in tank = {noOfGlass}")
```

```python
# Write a Pandas program to compare the elements of the two Pandas Series.

import pandas as pd

s1 = pd.Series([2, 4, 6, 8, 10])
s2 = pd.Series([1, 3, 5, 7, 10])

# to campare greater then 
pd.Series(s1 > s2)

# to campare less then 
pd.Series(s1 < s2)

# to campare equal to 
pd.Series(s1 == s2)
```

```python
# Write a function to change the data type of given a column or a Series. Function takes series and data type as input, returns the converted series. 

def data_type_change(ls,dtype):
    series = pd.Series(ls)
    converted_series = series.astype(dtype)
    return converted_series

ls = [1,2, 2.0, True, 100]
result = data_type_change(ls,float)
print(result)
    
```

```python
# Read `items.csv` making `item_name` as index. 
import pandas as pd
item = pd.read_csv('items.csv')
item_series = pd.Series(item['item_price'].values, index= item['item_name'])
item
```

```python
# Show no of nan values
item.isnull().sum()
```

```python
# Item price is given in $, so convert it to rupees without currency symbol.
item['item_price'] = item['item_price'].replace('[$]','',regex=True).astype(float)

usd_to_inr = 83
item['item_price'] = item['item_price'] * usd_to_inr
item['item_price'] = item['item_price'].round(2)
item
```

```python
item.value_counts().head(20).plot(kind='pie') 
```

```python
# Fill nan with mean of the series

item['item_price'] = item['item_price'].fillna(item['item_price'].mean())
```

```python
item['item_price'].isnull().sum()
```

```python
# Find mean price

item['item_price'].mean()
```
